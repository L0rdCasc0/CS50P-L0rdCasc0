 while the program is running.

```python
print("hello, world)
```

We can notice that we left out a quotation mark, running this snippet of code will produce an error, specifically a *"Syntax error"*.

**Runtime Errors**

Runtime errors refer to errors created by unexpected behaviours, for example asking for a numerical input and getting an alfanumerical input instead.

```python
x = int(input("What's x? "))
printf(f"x is {x}")
```

Our code should defend us from erroneous inputs implementing checks and balances to safeguard our code.

**Try**

The commands *try* and *except* are ways of testing user input before it can jeopardise our program:


```python
try:
    x = int(input("What's x? "))
    print(f"x is {x}")
except ValueError:
    print("x is not an integer")
```

This snippet of code prevents our program from outputting a ValueError, resulting in a custom error message for the user in case the input is not in the accepted format.

Another way of implementing this check would be:


```python
try:
    x = int(input("What's x? "))
except ValueError:
    print("x is not an integer")

print(f"x is {x}")
```

This "saves" some space which would make our code neater, the drawback is that we can incur in a new type of error: *NameError*.

Looking at the order of operations, it is possible that the code may try to assign an input as an integer when it is impossible to do so, resulting in a failure in the assignment of a value to the variable x, therefore no value can be printed at the end of the code.

**Else**

There is another way of implementing *try* to catch errors:

```python
try:
  x = int(input("What's x? "))
except ValueError:
  print("x is not an integer")
else:
  print(f"x is {x}")
```

If no exception occurs (so if it doesn't catch a ValueError) the code will run the *else* snippet of code.

We can improve our code implementing a loop to not simply end our code if our user inputs a wrong input:

```python
while True:
  try:
    x = int(input("What's x? "))
  except ValueError:
    print("x is not an integer")
  else:
    break

print(f"x is {x}")
```

The *while* loop will go on forever until the user inputs a "correct" input and the *else* break statement lets us out of the loop to run the last line of code.

**Creating a function to get an integer**

```python
def main():
  x = get_int()
  print(f"x is {x}")

def get_int():
  while True:
    try:
      x = int(input("What's x? "))
    except ValueError:
      print("x is not an integer")
    else:
      return x

main()
```

Substituting *break* for *return* in the else section we can give the x value to the main function while also breaking out of the loop.

It is also possible to rewrite in a more compact form, albeit more confusing to future reader.

```python
def main():
    x = get_int()
    print(f"x is {x}")

def get_int():
    while True:
        try:
            return int(input("What's x?"))
        except ValueError:
            print("x is not an integer")

main()
```

**Pass**

It is also possible to directly re-prompt the user for a correct input instead of printing an error statement every time:

```python
def main():
    x = get_int()
    print(f"x is {x}")

def get_int():
    while True:
        try:
            return int(input("What's x?"))
        except ValueError:
            pass

main()
```

The *pass* line lets us "skip" all the values that produce an error message, breaking out of the loop only when a "correct" input is inserted.
